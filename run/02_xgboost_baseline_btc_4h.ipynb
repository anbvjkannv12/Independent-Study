# BTCUSDT 4-hour baseline report

This notebook only reads the JSON created by the training CLI; it never fits a model.

```python
import json
from pathlib import Path

import pandas as pd

report_path = Path.cwd().parent / 'logs' / 'baseline_btc_4h.json'
if not report_path.exists():
    report_path = Path('logs/baseline_btc_4h.json')
report = json.loads(report_path.read_text(encoding='utf-8'))
report['task']
```

```python
pd.DataFrame(report['splits']).T.rename_axis('split')
```

```python
rows = []
for name, result in {'model': report['model'], **report['benchmarks']}.items():
    for split in ('validation', 'test'):
        metrics = result[split]
        rows.append({'approach': name, 'split': split, **{key: metrics[key] for key in ('accuracy', 'precision', 'recall', 'f1', 'roc_auc')}})
pd.DataFrame(rows).set_index(['split', 'approach']).sort_index()
```

```python
(pd.Series(report['model']['gain_importance'], name='gain')
 .sort_values(ascending=False)
 .rename_axis('feature')
 .to_frame())
```
