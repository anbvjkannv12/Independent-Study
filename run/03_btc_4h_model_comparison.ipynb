# BTCUSDT 4 小時三模型 Walk-forward 比較

本 notebook 僅讀取已儲存的 JSON 結果，不會進行訓練。

```python
import json
from pathlib import Path
import pandas as pd

report_path = Path('logs/btc_4h_model_comparison.json')
report = json.loads(report_path.read_text(encoding='utf-8'))
pd.DataFrame(report['folds'])
```

```python
rows = []
for model_name, model_report in report['models'].items():
    row = {'model': model_name}
    for metric, summary in model_report['test_aggregate'].items():
        row[f'{metric}_mean'] = summary['mean']
        row[f'{metric}_std'] = summary['sample_std']
    rows.append(row)
pd.DataFrame(rows).set_index('model').sort_values('roc_auc_mean', ascending=False)
```

```python
fold_rows = []
for model_name, model_report in report['models'].items():
    for fold_index, fold_report in enumerate(model_report['folds']):
        fold_rows.append({
            'model': model_name,
            'fold': fold_index,
            'threshold': fold_report['threshold'],
            'test_auc': fold_report['test']['uncalibrated_roc_auc'],
            'test_brier': fold_report['test']['brier_score'],
            'test_balanced_accuracy': fold_report['test']['balanced_accuracy'],
        })
pd.DataFrame(fold_rows).sort_values(['model', 'fold'])
```

結果未納入交易成本、滑價或持倉規則；請勿將此處的分類指標解讀為已證實的交易獲利。
