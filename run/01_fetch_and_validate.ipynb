```python
from pathlib import Path
import json
import subprocess
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'config.json').exists() or not (PROJECT_ROOT / 'scripts' / 'fetch_binance_5y.py').exists():
    raise RuntimeError('請從 C:/Users/user/專題 new (五年) 專案根目錄啟動 notebook')
print(PROJECT_ROOT)
```

```python
result = subprocess.run([sys.executable, 'scripts/fetch_binance_5y.py', '--dry-run', '--config', 'config.json'], check=True, capture_output=True, text=True)
print(result.stdout)
```

```python
RUN_FULL_DOWNLOAD = False
if RUN_FULL_DOWNLOAD:
    subprocess.run([sys.executable, 'scripts/fetch_binance_5y.py', '--config', 'config.json'], check=True)
else:
    print('目前為安全模式，尚未啟動完整下載。確認 dry-run 後再改為 True。')
```

```python
manifest_path = PROJECT_ROOT / 'logs' / 'fetch_manifest.json'
if manifest_path.exists():
    manifest = json.loads(manifest_path.read_text(encoding='utf-8'))
    for entry in manifest['symbols']:
        print(entry['symbol'], entry['status'], entry['row_count'], entry['first_open_time'], entry['last_open_time'])
else:
    print('尚未有 manifest，請先完成完整下載。')
```
